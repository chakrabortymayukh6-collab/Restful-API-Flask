- Question 1 : What is a RESTful API?

= The term REST API is used to refer to an API (Application Programming Interface) which conforms to REST (Representational State Transfer) principles, enabling applications to interact via HTTP.
Key concepts
- GET — Used to retrieve data.
- POST — Used for creating new data.
- PUT — Used to update/replaces data.
- DELETE — Used to delete data.
- Stateless — Every request includes all information required to fulfill it; no client state is retained by the server across multiple requests.
- Resources — Data can be accessed via URLs known as end-points.

```python
from flask import Flask, jsonify, request

app = Flask(__name__)

@app.route("/user", methods=["POST"])
def create_user():
    data = request.get_json()

    if not data or "name" not in data:
        return jsonify({"error": "name is required"}), 400

    user = {"id": 2, "name": data["name"], "role": data.get("role", "Analyst")}
    return jsonify(user), 201          # 201 = Created
```
